```python
gmsh.initialize()
gmsh.model.add("finfet3")
occ = gmsh.model.occ

Lsd, Lg, Wf, Hf, Tox, pitch, NF = 20, 20, 10, 30, 1, 40, 3
L = 2*Lsd + Lg

fins, shells = [], []
for i in range(NF):
    y0 = i*pitch
    fin = occ.addBox(0, y0, 0, L, Wf, Hf)
    ox  = occ.addBox(Lsd, y0-Tox, 0, Lg, Wf+2*Tox, Hf+Tox)
    shell, _ = occ.cut([(3, ox)], [(3, fin)], removeTool=False)
    fins.append(fin)
    shells += shell

occ.fragment([(3, f) for f in fins], shells)
occ.synchronize()
```

```python
bb = gmsh.model.getBoundingBox   # returns xmin, ymin, zmin, xmax, ymax, zmax

si_vols, ox_vols = [], []
for _, t in gmsh.model.getEntities(3):
    b = bb(3, t)
    (si_vols if (b[3]-b[0]) > Lg+1 else ox_vols).append(t)

def boundary(vols):
    return {t for v in vols
              for d, t in gmsh.model.getBoundary([(3, v)], oriented=False) if d == 2}

si_surf, ox_surf = boundary(si_vols), boundary(ox_vols)
iface = si_surf & ox_surf
tol = 1e-3
src  = [s for s in si_surf if abs(bb(2, s)[3]) < tol]        # faces at x = 0
drn  = [s for s in si_surf if abs(bb(2, s)[0] - L) < tol]    # faces at x = L
gate = [s for s in (ox_surf - iface)
        if (bb(2, s)[3] - bb(2, s)[0]) > tol                 # skip x-normal end caps
        and bb(2, s)[5] > tol]                               # skip the z = 0 bottom

pg = gmsh.model.addPhysicalGroup
pg(3, si_vols,      name="silicon")
pg(3, ox_vols,      name="oxide")
pg(2, src,          name="source")
pg(2, drn,          name="drain")
pg(2, gate,         name="gate")
pg(2, list(iface),  name="si_ox")

print(len(si_vols), len(ox_vols), len(src), len(drn), len(gate), len(iface))
```

```python
f = gmsh.model.mesh.field
f.add("Distance", 1)
f.setNumbers(1, "SurfacesList", list(iface))
f.setNumber(1, "Sampling", 50)
f.add("Threshold", 2)
f.setNumber(2, "InField", 1)
f.setNumber(2, "SizeMin", 1.0)    # start coarse; refine later
f.setNumber(2, "SizeMax", 4.0)
f.setNumber(2, "DistMin", 1.0)
f.setNumber(2, "DistMax", 8.0)
f.setAsBackgroundMesh(2)
for o in ("Mesh.MeshSizeFromPoints", "Mesh.MeshSizeFromCurvature",
          "Mesh.MeshSizeExtendFromBoundary"):
    gmsh.option.setNumber(o, 0)

gmsh.option.setNumber("Mesh.MshFileVersion", 2.2)
gmsh.model.mesh.generate(3)
gmsh.write("finfet3_nm.msh")
print(len(gmsh.model.mesh.getNodes()[0]), "nodes")
gmsh.finalize()
```

```python
def scale_msh(inp, out, s=1e-7):
    lines = open(inp).read().split("\n")
    i = lines.index("$Nodes"); n = int(lines[i+1])
    for k in range(i+2, i+2+n):
        idx, x, y, z = lines[k].split()
        lines[k] = f"{idx} {float(x)*s:.10e} {float(y)*s:.10e} {float(z)*s:.10e}"
    open(out, "w").write("\n".join(lines))

scale_msh("finfet3_nm.msh", "finfet3_cm.msh")
```

```python
from devsim import *
dev = "finfet3"
create_gmsh_mesh(mesh=dev, file="finfet3_cm.msh")
add_gmsh_region(mesh=dev, gmsh_name="silicon", region="si", material="Silicon")
add_gmsh_region(mesh=dev, gmsh_name="oxide",   region="ox", material="Oxide")
add_gmsh_contact(mesh=dev, gmsh_name="source", region="si", name="source", material="metal")
add_gmsh_contact(mesh=dev, gmsh_name="drain",  region="si", name="drain",  material="metal")
add_gmsh_contact(mesh=dev, gmsh_name="gate",   region="ox", name="gate",   material="metal")
add_gmsh_interface(mesh=dev, gmsh_name="si_ox", region0="si", region1="ox", name="si_ox")
finalize_mesh(mesh=dev)
create_device(mesh=dev, device=dev)

# sanity checks
xs = np.array(get_node_model_values(device=dev, region="si", name="x"))
print(xs.max())            # should be ~6e-6 (60 nm in cm)
print(get_region_list(device=dev), get_contact_list(device=dev), get_interface_list(device=dev))
```

```python
from devsim.python_packages.simple_physics import *
from devsim.python_packages.model_create import *

# Initial carrier densities from the solved potential
# (intrinsic density n_i and thermal voltage V_t are set by SetSiliconParameters)
CreateSolution(dev, "si", "Electrons")
CreateSolution(dev, "si", "Holes")
set_node_values(device=dev, region="si", name="Electrons", init_from="IntrinsicElectrons")
set_node_values(device=dev, region="si", name="Holes",     init_from="IntrinsicHoles")

# Now add transport
CreateSiliconDriftDiffusion(dev, "si")
for c in ("source", "drain"):
    CreateSiliconDriftDiffusionAtContact(dev, "si", c)
solve(type="dc", absolute_error=1e10, relative_error=1e-10, maximum_iterations=30)
```

```python
cur = {"source": 0.0, "drain": 0.0, "gate": 0.0}

def ramp(contact, target, step=0.05):
    n = max(1, int(round(abs(target - cur[contact]) / step)))
    for v in np.linspace(cur[contact], target, n + 1)[1:]:
        set_parameter(device=dev, name=GetContactBiasName(contact), value=v)
        solve(type="dc", absolute_error=1e10, relative_error=1e-10, maximum_iterations=30)
    cur[contact] = target

def Id():
    e = get_contact_current(device=dev, contact="drain", equation="ElectronContinuityEquation")
    h = get_contact_current(device=dev, contact="drain", equation="HoleContinuityEquation")
    return e + h

ramp("drain", 0.05)
print("Id at Vg=0, Vd=0.05:", Id())
ramp("gate", 0.5)
print("Id at Vg=0.5, Vd=0.05:", Id())
```

```python
def idvg(vd, vg_max=0.9, dvg=0.05):
    ramp("drain", vd, 0.05)
    out = []
    for vg in np.arange(0, vg_max + 1e-9, dvg):
        ramp("gate", vg, dvg)
        out.append((vg, abs(Id())))
        print(f"Vd={vd}  Vg={vg:.2f}  Id={out[-1][1]:.3e}")
    ramp("gate", 0.0, 0.1)
    return np.array(out)

lin = idvg(0.05)
sat = idvg(0.7)
```

```python
def vt_cc(d, icc=1e-6):
    # constant-current Vt; 1e-6 A is about 1e-7 A x (W/L) for W~210 nm, L=20 nm
    return np.interp(np.log10(icc), np.log10(d[:,1]), d[:,0])

def ss_min(d, vg_max=0.35):
    m = d[:,0] <= vg_max
    vg, i = d[m,0], np.log10(d[m,1])
    return (np.diff(vg) / np.diff(i) * 1000).min()   # mV/dec

vd_sat = 0.4   # use the Vd you actually ran
vt_lin, vt_sat = vt_cc(lin), vt_cc(sat)

print(f"Vt (lin)   : {vt_lin:.3f} V")
print(f"Vt (sat)   : {vt_sat:.3f} V")
print(f"SS (lin)   : {ss_min(lin):.1f} mV/dec")
print(f"DIBL       : {(vt_lin - vt_sat)/(vd_sat - 0.05)*1000:.0f} mV/V")
print(f"Ion/Ioff   : {lin[-1,1]/lin[0,1]:.2e}  (lin, Vg 0.9 V / 0 V)")
```

```python
def build_device(name, pocket_s=False, pocket_d=False, Np=3e18, sig=4e-7):
    create_gmsh_mesh(mesh=name, file="finfet3_cm.msh")
    add_gmsh_region(mesh=name, gmsh_name="silicon", region="si", material="Silicon")
    add_gmsh_region(mesh=name, gmsh_name="oxide",   region="ox", material="Oxide")
    add_gmsh_contact(mesh=name, gmsh_name="source", region="si", name="source", material="metal")
    add_gmsh_contact(mesh=name, gmsh_name="drain",  region="si", name="drain",  material="metal")
    add_gmsh_contact(mesh=name, gmsh_name="gate",   region="ox", name="gate",   material="metal")
    add_gmsh_interface(mesh=name, gmsh_name="si_ox", region0="si", region1="ox", name="si_ox")
    finalize_mesh(mesh=name)
    create_device(mesh=name, device=name)

    xs_j, xd_j, ls = 20e-7, 40e-7, 2e-7
    xs_p, xd_p = xs_j + 2e-7, xd_j - 2e-7      # pocket peaks 2 nm inside the channel
    acc = "1e15"
    if pocket_s: acc += f" + {Np}*exp(-((x-{xs_p})/{sig})^2)"
    if pocket_d: acc += f" + {Np}*exp(-((x-{xd_p})/{sig})^2)"
    node_model(device=name, region="si", name="Acceptors", equation=acc)
    node_model(device=name, region="si", name="Donors",
        equation=f"0.5e20*(1-tanh((x-{xs_j})/{ls})) + 0.5e20*(1+tanh((x-{xd_j})/{ls}))")
    node_model(device=name, region="si", name="NetDoping", equation="Donors-Acceptors")

    SetSiliconParameters(name, "si", 300)
    SetOxideParameters(name, "ox", 300)
    CreateSiliconPotentialOnly(name, "si")
    CreateOxidePotentialOnly(name, "ox", "default")
    for c in ("source", "drain"):
        set_parameter(device=name, name=GetContactBiasName(c), value=0.0)
        CreateSiliconPotentialOnlyContact(name, "si", c)
    set_parameter(device=name, name=GetContactBiasName("gate"), value=0.0)
    CreateOxideContact(name, "ox", "gate")
    CreateSiliconOxideInterface(name, "si_ox")
    solve(type="dc", absolute_error=1.0, relative_error=1e-10, maximum_iterations=50)

    CreateSolution(name, "si", "Electrons")
    CreateSolution(name, "si", "Holes")
    set_node_values(device=name, region="si", name="Electrons", init_from="IntrinsicElectrons")
    set_node_values(device=name, region="si", name="Holes",     init_from="IntrinsicHoles")
    CreateSiliconDriftDiffusion(name, "si")
    for c in ("source", "drain"):
        CreateSiliconDriftDiffusionAtContact(name, "si", c)
    solve(type="dc", absolute_error=1e10, relative_error=1e-8, maximum_iterations=40)
```
